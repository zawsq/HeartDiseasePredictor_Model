```python
#CHECKS FOR THE AMOUNT OF INCONSISTENCIES AND USE THIS AGAIN AFTER IMPLEMENTING THE CLEANING TO CHECK AGAIN
# Renaming columns to match with df2 and df3
df1_cleaned = df1_cleaned.rename(columns={"GenHealth": "GeneralHealth", "MentalHealth": "MentalHealthDays", "PhysicalHealth": "PhysicalHealthDays"})

df1_check1 = df1_cleaned[
    (df1_cleaned['GeneralHealth'].isin(["Very good", "Excellent"])) & (
        (df1_cleaned['KidneyDisease'] == 'Yes') |
        (df1_cleaned['Asthma'] == 'Yes') |
        (df1_cleaned['SkinCancer'] == 'Yes') |
        (df1_cleaned['Diabetic'] == 'Yes')
    )
]

df1_check1.shape
```

```python

def determine_genhealth(row):
    """
    Determines 'GeneralHealth' based on chronic conditions and physical/mental health days.
    """
    # Check for high PhysicalHealthDays and MentalHealthDays scores
    if row['PhysicalHealthDays'] >= 25 or row['MentalHealthDays'] >= 25:
        return 'Poor'  # Assign 'Poor' if either score is high
    
    # List of chronic conditions to check for
    chronic_conditions = [
        row['Asthma'],
        row['SkinCancer'],
        row['KidneyDisease'],
        row['Diabetic'],
        row['HeartDisease']
    ]
    
    # Count the number of chronic conditions present
    conditions_count = chronic_conditions.count('Yes')
    
    # Assign 'GeneralHealth' based on the number of chronic conditions
    if conditions_count >= 3:  # Three or more conditions
        return 'Poor'
    elif conditions_count == 2:  # Two conditions
        return 'Fair'
    elif conditions_count == 1:  # One condition
        return 'Good'
    
    # If no chronic conditions, return existing GeneralHealth value
    return row['GeneralHealth']

# Apply the function to each row
df1_cleaned['GeneralHealth'] = df1_cleaned.apply(determine_genhealth, axis=1)
```

## **CHECKING FOR OUTLIERS AND PERCENTAGE**

```python
# Identify outliers
def calculate_outlier_percentage(df):
    outlier_percentages = {}
    
    for column in df.select_dtypes(include='number').columns:
       
        Q1 = df[column].quantile(0.25)
        Q3 = df[column].quantile(0.75)
        IQR = Q3 - Q1
        
        
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        
        
        outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
        outlier_percentage = len(outliers) / len(df) * 100
        
        outlier_percentages[column] = outlier_percentage
    
    return outlier_percentages


datasets = {
    "dataset_1": df1_cleaned,
    "dataset_2": df2_cleaned,
    "dataset_3": df3_cleaned
}
#loop through the dataset and calculate the outliers in columns
for i, k in datasets.items():
    outlier_percentage = calculate_outlier_percentage(k)
    print(f"{i}: OUTLIER PERCENTAGE {outlier_percentage}\n")
```

## **From here we are gonna do train test split so that we avoid dataleak when we are doing mice imputation , the test set will remain to be the unknown real world data**

Heart disease can be imputed but since im trying to avoid making bias as much as possible its better that we just drop the missing values on our target variable and impute the other features that way we avoid making the entire missing values in y as ='No' since its imbalanced and majority are 'no'(its okay to drop since the amount of missing values in our heart disease is just 3k a small portion of the dataset)

We will train (X_train first learn from the data) then use that model to impute the x_test using (.transform) the model will impute the values without it learning from the test set so we avoid data leaks the same thing will be done for y_train and y_test

train_test_split(stratify=y ensures that both the training and test sets have the same proportions of each class as the original dataset.) 
since we just saw from the distribution plots that our target variable 'HeartDisease' is an imbalanced class meaning that the amount of no is comparatively much more overwhelming than the 'yes' values this can introduce bias so such we will use techniques to balance the dataset and much more later

```python
"""NOTE: that running this cell may take a while(expected time 100 seconds)"""
from sklearn.model_selection import train_test_split
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer


# Drop rows on our missing target (HeartDisease) values in df3_cleaned
df3_cleaned = df3_cleaned.dropna(subset=['HeartDisease'])

#EXCLUDING THE TWO FEATURES WE TALKED ABOUT
X_df3 = df3_cleaned.drop(columns=['HeartDisease', 'State', 'RaceEthnicityCategory'])
y_df3 = df3_cleaned['HeartDisease']

# 2. Train-test split with stratification
X_train_df3, X_test_df3, y_train_df3, y_test_df3 = train_test_split(X_df3, y_df3, test_size=0.2, random_state=42, stratify=y_df3)

#Initiating mice
mice_imputer = IterativeImputer(max_iter=50, random_state=42)

#train only on xtrain
X_train_df3 = pd.DataFrame(mice_imputer.fit_transform(X_train_df3), columns=X_train_df3.columns)
#then apply transform on x_test
X_test_df3 = pd.DataFrame(mice_imputer.transform(X_test_df3), columns=X_test_df3.columns)

```

+ ### **Now next is since we have encoded the data in order to use MICE and use MICE to impute the missing value now we can go back to what we talked about earlier which is fixing the inconsistent data that we might have on our now combined dataset 2 and 3 on the GeneralHealth Feature where the values have 'excellent' for example but has 4 chronic diseases(Stroke,heart disease,kidneydisease, etc) and physical and mental health scores poeple with more than 25 as poor**

We will just use the same function we did earlier but we modified it a bit to account for the feature encoding and other chronic diseases we have in our combined dataset 2 and 3 which has more features than dataset 1

```python
X_train_combined.columns
```

```python
def determine_genhealth(row):
    
    '''Check for high Physical and Mental Health scores Classifying individuals with 25 or more days of health problems as 
    'Poor' is reasonable because it indicates a significant impact on their daily lives.'''
    
    if row['PhysicalHealthDays'] >= 25 or row['MentalHealthDays'] >= 25:
        return 1  

    #chronic diseases to check(0 for No and 1 for Yes) not including heart disease because that is our target variable
    chronic_conditions = [
        row['HadAsthma'],
        row['HadSkinCancer'],
        row['HadCOPD'],
        row['HadDepressiveDisorder'],
        row['HadKidneyDisease'],
        row['HadArthritis'],
        row['HadDiabetes'],
        row['HadAngina'],
        row['HadStroke'],
    ]
    
    # Count how many chronic conditions are present
    conditions_count = sum(chronic_conditions) 




    
    #now filter the inconsistencies to respective values
    if conditions_count >= 3:  # Three or more conditions
        return 0  # 'Poor'
    elif conditions_count == 2:  # Two conditions
        return 1  # 'Fair'
    elif conditions_count == 1:  # One condition
        return 2  # 'Good'

    
    # If the existing GenHealth value is 'Poor', because we cant have poor turn into 'good' because it has only 1 chronic condition example
    existing_genhealth = row['GeneralHealth']
    if existing_genhealth == 1:  
        return 1  
    
    # Default case: return existing GenHealth value
    return existing_genhealth


# Apply the function to transform the GenHealth values in the combined dataset xtrain and xtest
X_train_combined['GeneralHealth'] = X_train_combined.apply(determine_genhealth, axis=1)
X_test_combined['GeneralHealth'] = X_test_combined.apply(determine_genhealth, axis=1)
```

## **FEATURE INTERACTION**

 + **Addition**: Use addition when you want to combine features without amplifying their effects on each other. This method is useful when the combined value should represent the accumulation of individual risks or factors. For example, in a lifestyle_score, adding smoking and physical inactivity gives a cumulative measure of unhealthy behaviors, where each behavior has an independent contribution to the total score.

+ **Multiplication**: Use multiplication when you want the interaction to reflect a synergistic effect (where the presence of both features increases the risk more than either alone). Multiplication is suitable when you believe that certain features combined have a stronger or nonlinear impact than they do individually. This is often used with continuous variables, but it can also be used with ordinal variables if the interaction effect makes sense for your problem.

+ **Here as we saw earlier in our data analysis that a combined synergy of chronic disease greatly increases your risk of heart disease such as we will do feature interaction on those features (note: it has plus 1 because if the value of the feature is 0 it will become 0 thus we add +1)**


+ **lifestyle score is made through by combining other risk factors that might influence heart disease combining all of them into a single feature rather than treating them as an individual feature affecting heart disease**

+ **another thing we could add is combining physical and mental health days to make a healthdays score same thing we are capturing the combined  variable which could be predictive of heart disease**

+ **Most of whats used here and combined are features that have some correlation to heart disease as we saw in the correlation heatmap and feature importance score (by random forest) All of these new features are not final we can either drop them if they serve no correlation so we'll check now again using our random forest selector creating new features like this is important when we want the model to capture more relationship of our features that the model might miss and it also reduces dimentionality**

+ **There are some other possible feature interaction or feature creation we can do but for now we'll leave it here feel free to do so if you want**

```python

# Multiplicative score for compounded risk
X_train_combined['chronic_disease_score'] = (
    (X_train_combined['HadStroke'] + 1) * (X_train_combined['HadCOPD'] + 1) * (X_train_combined['HadDiabetes'] + 1)
    * (X_train_combined['HadArthritis'] + 1) * (X_train_combined['HadKidneyDisease'] + 1) * (X_train_combined['HadSkinCancer'] + 1) * (X_train_combined['HadAsthma'] + 1)
    
                                                                                                        
)

#additive
X_train_combined['health_days'] = X_train_combined['PhysicalHealthDays'] + X_train_combined['MentalHealthDays']

#making a lifestyle feature based on the features included below
def calculate_lifestyle_score(row):
    score = 0
    
    # 1. Smoker Status
    if row['SmokerStatus'] == 3:  # Current smoker - now smokes every day
        score -= 1
    elif row['SmokerStatus'] == 2:  # Current smoker - smokes some days
        score -= 0.5
    elif row['SmokerStatus'] == 1:  # Former smoker
        score += 0  # Neutral for former smokers
    elif row['SmokerStatus'] == 0:  # Never smoked
        score += 1
    
    # 2. E-Cigarette Usage (Vaping)
    if row['ECigaretteUsage'] == 3:  # Use e-cigarettes every day
        score -= 1
    elif row['ECigaretteUsage'] == 2:  # Use e-cigarettes some days
        score -= 0.5
    elif row['ECigaretteUsage'] == 1:  # Never used e-cigarettes in my entire life
        score += 0  # Neutral for non-users
    elif row['ECigaretteUsage'] == 0:  # Not at all (right now)
        score += 1
    
    # 3. Alcohol Drinkers
    if row['AlcoholDrinkers'] == 1:  # Alcohol drinker
        score -= 1
    elif row['AlcoholDrinkers'] == 0:  # Non-alcohol drinker
        score += 1
    
    # 4. Physical Activity
    if row['PhysicalActivities'] == 0:  # No physical activity
        score -= 1
    elif row['PhysicalActivities'] == 1:  # Any physical activity
        score += 1
    
    # 5. Removed Teeth
    if row['RemovedTeeth'] == 0:  # No teeth removed
        score += 1
    elif 1 <= row['RemovedTeeth'] <= 1:  # 1 to 5 teeth removed
        score += 0  # Neutral
    elif 2 <= row['RemovedTeeth'] <= 2:  # 6 or more, but not all teeth removed
        score -= 0.5
    elif row['RemovedTeeth'] == 3:  # All teeth removed
        score -= 1
    
    # 6. Sleep Hours
    if row['SleepHours'] < 6:  # Less than 6 hours of sleep
        score -= 1
    elif row['SleepHours'] > 10:  # More than 10 hours of sleep
        score -= 1
    elif 7 <= row['SleepHours'] <= 9:  # Optimal sleep (7-9 hours)
        score += 1
    else:  # Between 6 and 7, or between 9 and 10 hours of sleep
        score += 0  # Neutral
    
    return score

# Apply and create the new feature lifestyle score
X_train_combined['lifestyle_score'] = X_train_combined.apply(calculate_lifestyle_score, axis=1)



#dropping the features we used because they will introduce noise or multicollineariy if you keep them while we have the feature interactions
#feel free to change the features or include them and rerun the feature selection model to test which scores the highest feature importance
X_train_combined.drop(['HadStroke', 'HadCOPD', 'HadDiabetes', 'HadArthritis', 'PhysicalHealthDays', 'MentalHealthDays',
                      'ECigaretteUsage', 'SmokerStatus', 'AlcoholDrinkers', 'PhysicalActivities', 'MentalHealthDays', 'SleepHours' 
                     ,'HadAsthma' , 'HadSkinCancer' , 'HadKidneyDisease', 'RemovedTeeth','HighRiskLastYear','TetanusLast10Tdap'
                     ], axis=1, inplace=True )
```

```python
#(this cell might take a few mins to run :) )
from sklearn.feature_selection import SelectFromModel
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.metrics import make_scorer, f1_score

# ReInitialize and fit Random Forest model so that we avoid making biases lets just call it rf_model_2
rf_model_2 = RandomForestClassifier(random_state=42)
rf_model_2.fit(X_train_combined, y_train_combined)

# Feature selection based on feature importance
selector = SelectFromModel(rf_model_2, prefit=True)


# Transform training and test sets to include only selected features
X_train_selected = selector.transform(X_train_combined)
X_test_selected = selector.transform(X_test_combined)

# Convert selected features back to DataFrame for clarity
selected_features = X_train_combined.columns[selector.get_support()]
X_train_selected = pd.DataFrame(X_train_selected, columns=selected_features)
X_test_selected = pd.DataFrame(X_test_selected, columns=selected_features)

# Initialize Stratified K-Folds for cross-validation
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# Use F1 score as the evaluation metric for cross-validation
f1_scorer = make_scorer(f1_score)

# Perform cross-validation with selected features
f1_scores = cross_val_score(rf_model_2, X_train_selected, y_train_combined, cv=cv, scoring=f1_scorer)

# Output cross-validation F1 scores
print(f"F1 Cross-validated scores with selected features: {f1_scores}")
print(f"Mean F1 score with selected features: {f1_scores.mean()}")

# Display selected features and shape of the transformed data
print("Selected features:", selected_features)
print("Shape of selected data:", X_train_selected.shape)
```
